```python
# Step 1: Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visual aesthetic configuration
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (14, 7)
plt.rcParams["font.sans-serif"] = "DejaVu Sans"
print("Libraries imported successfully!")
```

```python
# Step 2: Load and Inspect Dataset
df = pd.read_csv("unemployment_data.csv")
print("Dataset Shape:", df.shape)
print("\nColumn Data Types & Info:")
df.info()
print("\nFirst 5 Rows:")
df.head()
```

```python
# Step 3: Data Cleaning & Preprocessing
# Clean column header spaces
df.columns = df.columns.str.strip()

# Verify missing values
print("Missing Values per Column:")
print(df.isnull().sum())

# Drop missing if any
df = df.dropna()

# Parse datetime
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.strftime("%b")
print("Preprocessed successfully. Cleaned records:", len(df))
```

```python
# Step 4: Exploratory Data Analysis (EDA)
print("Statistical Summary:")
display(df.describe().T)

# Regional Unemployment ranking
regional_agg = df.groupby("Region")["Estimated Unemployment Rate (%)"].agg(["mean", "median", "max", "std"])
regional_agg = regional_agg.sort_values(by="mean", ascending=False)
display(regional_agg)
```

```python
# Step 5: COVID-19 Shock Visualizations
# 1. Timeline Trend
plt.figure(figsize=(14, 6))
trend = df.groupby("Date")["Estimated Unemployment Rate (%)"].mean().reset_index()
sns.lineplot(data=trend, x="Date", y="Estimated Unemployment Rate (%)", color="#0284c7", linewidth=2.5, marker="o")
plt.axvline(pd.to_datetime("2020-03-24"), color="#dc2626", linestyle="--", label="Lockdown Enacted (Mar 2020)")
plt.title("National Unemployment Rate Trend: COVID-19 Lockdown Surge", fontsize=16, fontweight="bold")
plt.ylabel("Unemployment Rate (%)")
plt.legend()
plt.show()

# 2. Regional Bar Chart
plt.figure(figsize=(15, 7))
order = df.groupby("Region")["Estimated Unemployment Rate (%)"].mean().sort_values(ascending=False).index
sns.barplot(data=df, x="Region", y="Estimated Unemployment Rate (%)", order=order, palette="mako")
plt.xticks(rotation=90)
plt.title("Average Unemployment Rate by Region/State", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()

# 3. Monthly Heatmap
pivot_tab = df.pivot_table(values="Estimated Unemployment Rate (%)", index="Year", columns="Month", aggfunc="mean")
plt.figure(figsize=(12, 5))
sns.heatmap(pivot_tab, annot=True, fmt=".1f", cmap="YlOrRd", cbar_kws={"label": "Unemployment (%)"})
plt.title("Seasonality & COVID-19 Shock Heatmap (Year x Month)", fontsize=15, fontweight="bold")
plt.show()
```
